```python
import sys
import importlib
from pathlib import Path

ANCHO = 64
print("═" * ANCHO)
print("  VERIFICACIÓN DEL ENTORNO".center(ANCHO))
print("═" * ANCHO)

fallas = []

# ── 1 · Versión de Python ───────────────────────────────────────────────
v = sys.version_info
if v >= (3, 11):
    print(f"  ✅  Python {v.major}.{v.minor}.{v.micro}")
else:
    print(f"  ❌  Python {v.major}.{v.minor} — se necesita 3.11 o superior")
    fallas.append("Paso 2 de la guía de instalación")

# ── 2 · ¿Estamos en el entorno virtual del curso? ───────────────────────
en_venv = sys.prefix != sys.base_prefix
ruta_venv = Path(sys.prefix).name
if en_venv and ruta_venv == ".venv":
    print(f"  ✅  Entorno virtual activo (.venv)")
elif en_venv:
    print(f"  ⚠️   Entorno virtual activo, pero se llama '{ruta_venv}'")
    print(f"      Puede ser de otro proyecto. Revisa el kernel seleccionado.")
    fallas.append("Problema 3 · VS Code no encuentra el kernel")
else:
    print("  ❌  NO estás en un entorno virtual")
    print("      El kernel apunta al Python del sistema, no al del curso.")
    fallas.append("Problema 3 · VS Code no encuentra el kernel")

# ── 3 · Librerías ───────────────────────────────────────────────────────
print("─" * ANCHO)
for paquete, minimo in [("pandas", "3.0"), ("duckdb", "1.5"),
                        ("matplotlib", "3.9"), ("pyarrow", "20.0")]:
    try:
        mod = importlib.import_module(paquete)
        print(f"  ✅  {paquete:<12} {getattr(mod, '__version__', '?')}")
    except ImportError:
        print(f"  ❌  {paquete:<12} no instalado")
        fallas.append("Paso 4 · pip install -r requirements.txt")

# ── 4 · Datos del curso ─────────────────────────────────────────────────
print("─" * ANCHO)
raiz = Path.cwd()
if raiz.name == "notebooks":
    raiz = raiz.parent
esperados = [
    "datos/crudos/hechos_gasto.csv",
    "datos/crudos/presupuesto.csv",
    "datos/crudos/dim_centro_costo.csv",
    "utils/verificar.py",
]
for rel in esperados:
    if (raiz / rel).exists():
        print(f"  ✅  {rel}")
    else:
        print(f"  ❌  {rel} no encontrado")
        fallas.append("Paso 1 · descargar el curso completo")

# ── 5 · El módulo de autoverificación importa ───────────────────────────
print("─" * ANCHO)
try:
    sys.path.insert(0, str(raiz))
    from utils.verificar import verificar          # noqa: F401
    print("  ✅  utils.verificar importa correctamente")
except Exception as e:
    print(f"  ❌  no se pudo importar utils.verificar: {e}")
    fallas.append("Paso 1 · descargar el curso completo")

# ── Veredicto ───────────────────────────────────────────────────────────
print("═" * ANCHO)
if not fallas:
    print("  🎉  TODO LISTO — abre 01_caso_e_ingesta.ipynb")
else:
    print(f"  Hay {len(fallas)} cosa(s) por resolver. Ve a docs/01_INSTALACION.md:")
    for f in dict.fromkeys(fallas):
        print(f"      → {f}")
    print()
    print("  Si no lo resuelves en 5 minutos, avisa al instructor y sigue:")
    print("  puedes incorporarte al curso en cualquier bloque.")
print("═" * ANCHO)
```
